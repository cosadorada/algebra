# Buchberger's Algorithm
This is a walk-through example of division algorithms, and the use of Buchberger's algorithm, when we work in multivariate polynomial rings.

## Euclidean algorithm for $F[X]$
First of all, we demonstrate the relative ease we can do 'division' in these rings. For $F$ a field, we can demonstrate that $F[X]$ is a Euclidean domain. This means that we can, while looking to degrees of polynomials as our notion of 'size', for two polynomials $f$ and $g$, find polynomials $q$ and $r$ such that

\begin{equation}
f = g \cdot q + r
\end{equation}
where $\text{deg}(r)< \text{deg}(g)$.

This algorithm (repeated) also demonstrates that $F[X]$ is a Principal Ideal Domain (PID). Note the direct analogy to the division and Euclidean algorithms in $\mathbb{Z}$. This means it is fairly easy to determine to determine whether a given polynomial lies inside an ideal.

## The problem

We run into issues when we extend into multiple variables and consider $F[X, Y]$. Gauss's Lemma ensures that $F[X,Y]$ is a Unique Factorisation Domain (UFD). However, it is certainly not a PID, taking for example the ideal $(X, Y)$. How should we tell whether a polynomial $f$ lies inside our ideal $(f_1, f_2, \dots, f_n)$?

We might ask whether we can find a way to find for a given ideal, a 'useful generating set' or 'basis', whatever that might look like. It turns out (thanks to Buchberger) that by imposing an 'ordering' of the variables in the ring, we can answer this ideal membership problem. 

We should also note that Hilbert's Basis Theorem guarantees that since $F$ is Noetherian, then $F[X]$ is also Noetherian (and thus $F[X, Y]$ and so on). This means that any ideal is finitely generated, so the question is a sensible one as well. 

## Ordering 

We may go with the 'lexicographic' ordering in our example, meaning that $X_1^aX_2^b > X_1^cX_2^d$ if $a>c$, or $a=c$ and $b>d$. (Similarly for more variables) Through this, we can give the 'leading term' of a polynomial, according to this ordering. There are other choices of ordering available, but this one was simple to implement (since Python tuple comparison already does this automatically).

Equipped with our leading term, our aim is essentially to, looking at leading terms, 'divide' $f$ by the set $\{f_1, \dots f_n\}$, i.e. writing it $f = f_1 \cdot q_1 + \dots + f_n \cdot q_n + r$, where $r$ has no terms divisible by the leading terms of the $f_i$. Hopefully, $r=0$ is equivalent to $f$ belonging to the ideal. The division described is our 'multivariate division algorithm' (with respect to the ordering chosen).

This is, unfortunately, not the case. Take the case $(x^2-y, xy - 1)$ and $f=x-y^2$. $f$ is not 'divisible' by either of our generators, comparing 'size' or lexicographic order of our leading terms - so we would be inclined to say that the remainder $r$ was non-zero. However, we can check that 
\begin{equation} f = y \cdot (x^2-y) - x\cdot (xy-1)
\end{equation}
The issue that we come across, is that leading terms can cancel each other out. For our 'basis' to be useful in solving our problem, we should have that the leading term of non-zero $f$, $LT(f)$ is divisible by at least one of the basis elements' leading terms. This is known as a Gröbner basis.

## The Buchberger algorithm
Thanks to Buchberger, we have a way to obtain these Gröbner bases (and so we know also that they exist). We define the S-polynomial of two polynomials $f$ and $g$ as:

\begin{equation} S(f, g) := \frac{lcm(LT(f), LT(g))}{LT(f)}\cdot f - \frac{lcm(LT(f), LT(g))}{LT(g)}\cdot g
\end{equation}
Note that this cancels the leading terms, which was our main issue.

The algorithm proceeds as follows:

- We start with a generating set $G = \{ f_1, f_2, \dots\}$.
- Then for each pair $f$, $g$ in $G$, compute the S-polynomial $S(f, g)$. 
- Divide $S(f, g)$ by $G$; if remainder is non-zero, add it to $G$.
- Repeat until all S-polynomials are 0.

This yields our Gröbner basis, and will terminate thanks to Hilbert's Basis Theorem (applied to the ascending chain of ideals generated by leading terms).

It is then a theorem that for an ideal $I$ and Gröbner basis $G$, $f \in I$ if and only if $f$ gives remainder 0 when 'divided' by $G$.
